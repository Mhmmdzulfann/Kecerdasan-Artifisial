## **1. Eksplorasi Bawah Permukaan Menggunakan Metode Geolistrik ERT**

Eksplorasi bawah permukaan dilakukan pada suatu daerah untuk mengetahui variasi kondisi material di bawah permukaan. Metode Electrical Resistivity Tomography (ERT) digunakan untuk mengidentifikasi perbedaan resistivitas material sehingga dapat membantu mengetahui struktur dan zona anomali bawah permukaan. Hasil survei digunakan sebagai dasar untuk menentukan kondisi geologi bawah permukaan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode Electrical Resistivity Tomography (ERT).

T — Task: Identifikasi variasi resistivitas bawah permukaan berdasarkan hasil pengukuran ERT dan buat penampang resistivitas 2D.

C — Context: Suatu wilayah membutuhkan informasi mengenai kondisi bawah permukaan untuk mengetahui adanya perbedaan material atau zona anomali.

I — Input: Data posisi elektroda, konfigurasi elektroda, jarak antar-elektroda, dan nilai resistivitas hasil pengukuran.

K — Constraints: Jangan mengubah data pengukuran asli dan pertahankan satuan serta posisi elektroda.

O — Output: Hasil berupa penampang resistivitas 2D dan interpretasi zona resistivitas bawah permukaan.

Q — Quality Criteria: Penampang memiliki skala, legenda, satuan yang jelas, dan hasil interpretasi sesuai dengan pola resistivitas.

V — Optional Verification: Periksa kembali posisi elektroda, jumlah data, dan rentang nilai resistivitas.

Data sintetis: lintasan geolistrik sepanjang 100 m dengan kedalaman model 20 m, jarak elektroda 5 m, dan variasi resistivitas untuk menggambarkan perbedaan material bawah permukaan.
Tujuan: menghasilkan visualisasi penampang resistivitas 2D untuk mengidentifikasi zona resistivitas rendah, sedang, dan tinggi serta membantu interpretasi kondisi geologi bawah permukaan

## **2. Identifikasi Lapisan Bawah Permukaan Menggunakan Metode Seismik Refraksi**

Studi bawah permukaan dilakukan untuk mengetahui struktur lapisan batuan berdasarkan karakteristik gelombang seismik. Metode seismik refraksi digunakan dengan menganalisis waktu tiba gelombang P pada beberapa geophone. Hasil pengolahan digunakan untuk memperkirakan kecepatan gelombang dan kedalaman batas antar lapisan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode seismik refraksi.

T — Task: Identifikasi kecepatan gelombang dan kedalaman lapisan bawah permukaan berdasarkan data waktu tiba gelombang P.

C — Context: Suatu wilayah membutuhkan informasi mengenai struktur lapisan bawah permukaan dangkal untuk mendukung interpretasi kondisi geologi.

I — Input: Data posisi geophone, jarak sumber–geophone, dan waktu tiba gelombang P.

K — Constraints: Jangan mengubah data waktu tiba asli dan gunakan satuan yang konsisten.

O — Output: Grafik travel time, nilai kecepatan gelombang, estimasi kedalaman, dan model lapisan bawah permukaan.

Q — Quality Criteria: Grafik mudah dibaca, perhitungan kecepatan konsisten, dan estimasi kedalaman sesuai dengan model yang digunakan.

V — Optional Verification: Periksa kembali kemiringan grafik travel time dan hasil perhitungan kecepatan.

Data sintetis: lintasan sepanjang 100 m dengan 21 geophone berjarak 5 m dan model dua lapisan bawah permukaan. Data waktu tiba gelombang P dibuat untuk simulasi identifikasi kecepatan dan kedalaman lapisan.
Tujuan: menghasilkan grafik travel time, menghitung kecepatan gelombang P pada setiap lapisan, serta memperkirakan kedalaman batas lapisan menggunakan metode waktu intersep (intercept-time method).

## **3. Identifikasi Struktur Bawah Permukaan Menggunakan Metode Magnetotelurik (MT)**

Studi bawah permukaan dilakukan untuk mengetahui variasi sifat kelistrikan material di bawah permukaan. Metode Magnetotelurik (MT) digunakan dengan memanfaatkan variasi medan listrik dan medan magnet alami bumi. Hasil pengolahan data digunakan untuk memperoleh informasi resistivitas semu dan fase terhadap frekuensi.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode Magnetotelurik (MT).

T — Task: Analisis data medan listrik dan medan magnet untuk memperoleh kurva resistivitas semu dan fase terhadap frekuensi.

C — Context: Suatu daerah penelitian membutuhkan informasi mengenai variasi resistivitas bawah permukaan untuk membantu memahami kondisi geologi.

I — Input: Data time series medan listrik dan medan magnet, frekuensi sampling, serta data kalibrasi instrumen.

K — Constraints: Jangan mengubah data asli dan tandai data yang memiliki noise atau kurva tidak stabil.

O — Output: Kurva resistivitas semu dan fase terhadap frekuensi serta hasil pemeriksaan kualitas data.

Q — Quality Criteria: Kurva memiliki sumbu, satuan, dan legenda yang jelas serta data noise diberi tanda.

V — Optional Verification: Bandingkan hasil sebelum dan sesudah proses smoothing untuk memastikan tidak muncul anomali baru.

Data sintetis: simulasi medan listrik dan medan magnet pada beberapa frekuensi. Data ini digunakan untuk menghitung resistivitas semu dan fase, menandai data yang berpotensi mengandung noise, serta membandingkan kurva sebelum dan sesudah smoothing.
Tujuan: memperoleh kurva resistivitas semu (Ωm) dan fase (°) terhadap frekuensi (Hz), serta memeriksa kestabilan data.

## **4. Identifikasi Profil Kecepatan Gelombang Geser Menggunakan Metode MASW**

Studi bawah permukaan dilakukan untuk mengetahui variasi kecepatan gelombang geser terhadap kedalaman. Metode Multichannel Analysis of Surface Waves (MASW) digunakan dengan menganalisis gelombang permukaan yang direkam oleh beberapa geophone. Hasil survei digunakan untuk memperoleh profil Vs dan membantu mengetahui karakteristik material bawah permukaan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode Multichannel Analysis of Surface Waves (MASW).

T — Task: Identifikasi profil kecepatan gelombang geser (Vs) berdasarkan kurva dispersi data MASW.

C — Context: Suatu wilayah membutuhkan informasi mengenai variasi kecepatan gelombang geser terhadap kedalaman untuk mengetahui karakteristik bawah permukaan.

I — Input: Data rekaman seismik, posisi geophone, jarak antar-geophone, posisi sumber, dan waktu sampling.

K — Constraints: Jangan mengubah data rekaman asli dan pertahankan posisi serta jarak antar-geophone.

O — Output: Shot gather, kurva dispersi, dan profil Vs terhadap kedalaman.

Q — Quality Criteria: Kurva dispersi dapat dibaca dengan jelas, satuan konsisten, dan profil Vs dapat merepresentasikan data dispersi.

V — Optional Verification: Bandingkan kurva dispersi hasil pengolahan dengan data observasi dan periksa kesesuaian model Vs.

Data sintetis: simulasi rekaman seismik dari 24 geophone dengan jarak antar-geophone 2 m. Data digunakan untuk membuat shot gather, kurva dispersi, dan profil kecepatan gelombang geser (\(V_s\)) terhadap kedalaman.
Tujuan: memperlihatkan tahapan dasar analisis MASW untuk memperkirakan variasi kecepatan gelombang geser pada lapisan bawah permukaan.

## **5. Pemeriksaan Kualitas Data Menggunakan Metode GPR**

Survei Ground Penetrating Radar (GPR) dilakukan untuk memperoleh informasi mengenai kondisi bawah permukaan dangkal. Sebelum dilakukan pengolahan lebih lanjut, data perlu diperiksa untuk mengetahui kelengkapan dan kualitas hasil akuisisi. Pemeriksaan dilakukan terhadap jumlah trace, interval sampling, frekuensi antena, dan posisi lintasan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai pengolahan dan pemeriksaan data Ground Penetrating Radar (GPR).

T — Task: Periksa kualitas dan kelengkapan data GPR berdasarkan parameter akuisisi yang tersedia.

C — Context: Data GPR akan digunakan untuk analisis bawah permukaan sehingga perlu dipastikan memiliki kualitas dan kelengkapan yang memadai.

I — Input: Data trace GPR, interval sampling waktu, frekuensi antena, dan koordinat lintasan.

K — Constraints: Jangan mengubah raw trace, melakukan migrasi, atau menentukan velocity model.

O — Output: Laporan QC yang berisi jumlah trace, interval sampling, frekuensi antena, posisi, dan data yang bermasalah.

Q — Quality Criteria: Seluruh parameter utama tercatat, data bermasalah diberi flag, dan data asli tetap tidak berubah.

V — Optional Verification: Periksa kembali jumlah trace dan parameter sampling dengan metadata asli.

Data sintetis: simulasi data GPR dengan 200 trace, interval sampling 0,5 ns, frekuensi antena 200 MHz, dan posisi lintasan sepanjang 23 m.
Tujuan: memeriksa kelengkapan trace, interval sampling, frekuensi antena, koordinat lintasan, serta menandai data yang bermasalah tanpa mengubah data asli, melakukan migrasi, atau menentukan model kecepatan.
