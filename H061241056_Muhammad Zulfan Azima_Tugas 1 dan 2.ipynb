## **1. Eksplorasi Bawah Permukaan Menggunakan Metode Geolistrik ERT**

Eksplorasi bawah permukaan dilakukan pada suatu daerah untuk mengetahui variasi kondisi material di bawah permukaan. Metode Electrical Resistivity Tomography (ERT) digunakan untuk mengidentifikasi perbedaan resistivitas material sehingga dapat membantu mengetahui struktur dan zona anomali bawah permukaan. Hasil survei digunakan sebagai dasar untuk menentukan kondisi geologi bawah permukaan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode Electrical Resistivity Tomography (ERT).

T — Task: Identifikasi variasi resistivitas bawah permukaan berdasarkan hasil pengukuran ERT dan buat penampang resistivitas 2D.

C — Context: Suatu wilayah membutuhkan informasi mengenai kondisi bawah permukaan untuk mengetahui adanya perbedaan material atau zona anomali.

I — Input: Data posisi elektroda, konfigurasi elektroda, jarak antar-elektroda, dan nilai resistivitas hasil pengukuran.

K — Constraints: Jangan mengubah data pengukuran asli dan pertahankan satuan serta posisi elektroda.

O — Output: Hasil berupa penampang resistivitas 2D dan interpretasi zona resistivitas bawah permukaan.

Q — Quality Criteria: Penampang memiliki skala, legenda, satuan yang jelas, dan hasil interpretasi sesuai dengan pola resistivitas.

V — Optional Verification: Periksa kembali posisi elektroda, jumlah data, dan rentang nilai resistivitas.

Data sintetis: lintasan geolistrik sepanjang 100 m dengan kedalaman model 20 m, jarak elektroda 5 m, dan variasi resistivitas untuk menggambarkan perbedaan material bawah permukaan.
Tujuan: menghasilkan visualisasi penampang resistivitas 2D untuk mengidentifikasi zona resistivitas rendah, sedang, dan tinggi serta membantu interpretasi kondisi geologi bawah permukaan

```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm

# ============================================================
# 1. PEMBUATAN MODEL RESISTIVITAS ERT SINTETIS
# ============================================================

x = np.linspace(0, 100, 101)  # Jarak lintasan (m)
z = np.linspace(0, 20, 81)    # Kedalaman (m)
X, Z = np.meshgrid(x, z)

# Resistivitas latar belakang
rho = np.full_like(X, 100.0)  # ohm.m

# Zona resistivitas rendah di bagian tengah
rho[
    (X >= 35) & (X <= 60) &
    (Z >= 4) & (Z <= 12)
] = 15.0

# Zona resistivitas tinggi di bagian kanan bawah
rho[
    (X >= 70) & (Z >= 8)
] = 300.0

# ============================================================
# 2. INFORMASI MODEL DAN POSISI ELEKTRODA
# ============================================================

jarak_elektroda = 5
elektroda = np.arange(0, 101, jarak_elektroda)

print("=" * 55)
print("METODE ERT - MODEL RESISTIVITAS SINTETIS")
print("=" * 55)
print(f"Panjang lintasan       : {x.max():.0f} m")
print(f"Kedalaman model        : {z.max():.0f} m")
print(f"Jarak antar-elektroda  : {jarak_elektroda} m")
print(f"Jumlah elektroda       : {len(elektroda)}")
print(f"Resistivitas minimum   : {rho.min():.1f} ohm.m")
print(f"Resistivitas maksimum  : {rho.max():.1f} ohm.m")
print("Jenis data             : Sintetis")
print("=" * 55)

# ============================================================
# 3. VISUALISASI PENAMPANG RESISTIVITAS 2D
# ============================================================

fig, ax = plt.subplots(figsize=(12, 6))

im = ax.pcolormesh(
    X, Z, rho,
    shading="auto",
    cmap="turbo",
    norm=LogNorm(vmin=10, vmax=300)
)

ax.invert_yaxis()

# Posisi elektroda di permukaan
ax.scatter(
    elektroda,
    np.zeros_like(elektroda),
    color="black",
    s=12,
    label="Posisi elektroda"
)

cbar = fig.colorbar(im, ax=ax, pad=0.02)
cbar.set_label("Resistivitas (ohm.m)")

ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman model (m)")
ax.set_title("Penampang Resistivitas ERT 2D Sintetis")
ax.set_xticks(np.arange(0, 101, 10))
ax.set_yticks(np.arange(0, 21, 2))
ax.legend(loc="lower right")

fig.tight_layout()

# ============================================================
# 4. SIMPAN DAN TAMPILKAN GRAFIK
# ============================================================

nama_file = "penampang_ERT_sintetis.png"

fig.savefig(
    nama_file,
    dpi=200,
    bbox_inches="tight"
)

plt.show()

```

## **2. Identifikasi Lapisan Bawah Permukaan Menggunakan Metode Seismik Refraksi**

Studi bawah permukaan dilakukan untuk mengetahui struktur lapisan batuan berdasarkan karakteristik gelombang seismik. Metode seismik refraksi digunakan dengan menganalisis waktu tiba gelombang P pada beberapa geophone. Hasil pengolahan digunakan untuk memperkirakan kecepatan gelombang dan kedalaman batas antar lapisan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode seismik refraksi.

T — Task: Identifikasi kecepatan gelombang dan kedalaman lapisan bawah permukaan berdasarkan data waktu tiba gelombang P.

C — Context: Suatu wilayah membutuhkan informasi mengenai struktur lapisan bawah permukaan dangkal untuk mendukung interpretasi kondisi geologi.

I — Input: Data posisi geophone, jarak sumber–geophone, dan waktu tiba gelombang P.

K — Constraints: Jangan mengubah data waktu tiba asli dan gunakan satuan yang konsisten.

O — Output: Grafik travel time, nilai kecepatan gelombang, estimasi kedalaman, dan model lapisan bawah permukaan.

Q — Quality Criteria: Grafik mudah dibaca, perhitungan kecepatan konsisten, dan estimasi kedalaman sesuai dengan model yang digunakan.

V — Optional Verification: Periksa kembali kemiringan grafik travel time dan hasil perhitungan kecepatan.

Data sintetis: lintasan sepanjang 100 m dengan 21 geophone berjarak 5 m dan model dua lapisan bawah permukaan. Data waktu tiba gelombang P dibuat untuk simulasi identifikasi kecepatan dan kedalaman lapisan.
Tujuan: menghasilkan grafik travel time, menghitung kecepatan gelombang P pada setiap lapisan, serta memperkirakan kedalaman batas lapisan menggunakan metode waktu intersep (intercept-time method).

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. PEMBUATAN DATA SEISMIK REFRAKSI SINTETIS
# ============================================================

x = np.arange(5, 101, 5)  # Posisi geophone (m)

# Parameter model dua lapisan
v1 = 500.0   # Kecepatan lapisan pertama (m/s)
v2 = 1800.0  # Kecepatan lapisan kedua (m/s)
h = 8.0      # Kedalaman lapisan pertama (m)

# Waktu tiba gelombang langsung
t_direct = x / v1

# Sudut kritis dan waktu intersep
sin_ic = v1 / v2
cos_ic = np.sqrt(1 - sin_ic**2)
t_intercept = 2 * h * cos_ic / v1

# Waktu tiba gelombang refraksi
t_refracted = x / v2 + t_intercept

# Waktu tiba sintetik: memilih waktu tercepat
t_arrival = np.minimum(t_direct, t_refracted)

# ============================================================
# 2. ESTIMASI KECEPATAN DARI GRAFIK TRAVEL TIME
# ============================================================

# Segmen gelombang langsung
mask_direct = x <= 15
slope1, intercept1 = np.polyfit(
    x[mask_direct], t_arrival[mask_direct], 1
)

# Segmen gelombang refraksi
mask_refracted = x >= 35
slope2, intercept2 = np.polyfit(
    x[mask_refracted], t_arrival[mask_refracted], 1
)

# Konversi waktu dari sekon ke milisekon
v1_estimasi = 1 / (slope1 * 1000)
v2_estimasi = 1 / (slope2 * 1000)

# Estimasi kedalaman dengan waktu intersep
t_intersep_estimasi = intercept2
h_estimasi = (
    t_intersep_estimasi * v1_estimasi * 1000
    / (2 * np.sqrt(1 - (v1_estimasi / v2_estimasi)**2))
)

# ============================================================
# 3. INFORMASI HASIL
# ============================================================

print("=" * 55)
print("ANALISIS SEISMIK REFRAKSI - DATA SINTETIS")
print("=" * 55)
print(f"Jumlah geophone          : {len(x)}")
print(f"Jarak antar-geophone     : 5 m")
print(f"Kecepatan lapisan 1      : {v1_estimasi:.2f} m/s")
print(f"Kecepatan lapisan 2      : {v2_estimasi:.2f} m/s")
print(f"Estimasi kedalaman batas : {h_estimasi:.2f} m")
print("=" * 55)

# ============================================================
# 4. VISUALISASI GRAFIK TRAVEL TIME
# ============================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    x, t_arrival * 1000,
    color="black",
    label="Waktu tiba sintetis"
)

plt.plot(
    x[mask_direct],
    (slope1 * x[mask_direct] + intercept1) * 1000,
    color="blue",
    label="Gelombang langsung"
)

plt.plot(
    x[mask_refracted],
    (slope2 * x[mask_refracted] + intercept2) * 1000,
    color="red",
    label="Gelombang refraksi"
)

plt.xlabel("Jarak sumber–geophone (m)")
plt.ylabel("Waktu tiba (ms)")
plt.title("Grafik Travel Time Seismik Refraksi")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# ============================================================
# 5. VISUALISASI MODEL LAPISAN BAWAH PERMUKAAN
# ============================================================

fig, ax = plt.subplots(figsize=(10, 4))

ax.fill_between(
    [0, 100], 0, h_estimasi,
    color="burlywood",
    label=f"Lapisan 1: {v1_estimasi:.0f} m/s"
)

ax.fill_between(
    [0, 100], h_estimasi, h_estimasi + 12,
    color="lightsteelblue",
    label=f"Lapisan 2: {v2_estimasi:.0f} m/s"
)

ax.axhline(
    h_estimasi,
    color="black",
    linestyle="--",
    label=f"Batas lapisan: {h_estimasi:.2f} m"
)

ax.set_xlim(0, 100)
ax.set_ylim(h_estimasi + 12, 0)
ax.set_xlabel("Jarak lintasan (m)")
ax.set_ylabel("Kedalaman (m)")
ax.set_title("Model Lapisan Bawah Permukaan")
ax.legend()
plt.tight_layout()
plt.show()
```

## **3. Identifikasi Struktur Bawah Permukaan Menggunakan Metode Magnetotelurik (MT)**

Studi bawah permukaan dilakukan untuk mengetahui variasi sifat kelistrikan material di bawah permukaan. Metode Magnetotelurik (MT) digunakan dengan memanfaatkan variasi medan listrik dan medan magnet alami bumi. Hasil pengolahan data digunakan untuk memperoleh informasi resistivitas semu dan fase terhadap frekuensi.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode Magnetotelurik (MT).

T — Task: Analisis data medan listrik dan medan magnet untuk memperoleh kurva resistivitas semu dan fase terhadap frekuensi.

C — Context: Suatu daerah penelitian membutuhkan informasi mengenai variasi resistivitas bawah permukaan untuk membantu memahami kondisi geologi.

I — Input: Data time series medan listrik dan medan magnet, frekuensi sampling, serta data kalibrasi instrumen.

K — Constraints: Jangan mengubah data asli dan tandai data yang memiliki noise atau kurva tidak stabil.

O — Output: Kurva resistivitas semu dan fase terhadap frekuensi serta hasil pemeriksaan kualitas data.

Q — Quality Criteria: Kurva memiliki sumbu, satuan, dan legenda yang jelas serta data noise diberi tanda.

V — Optional Verification: Bandingkan hasil sebelum dan sesudah proses smoothing untuk memastikan tidak muncul anomali baru.

Data sintetis: simulasi medan listrik dan medan magnet pada beberapa frekuensi. Data ini digunakan untuk menghitung resistivitas semu dan fase, menandai data yang berpotensi mengandung noise, serta membandingkan kurva sebelum dan sesudah smoothing.
Tujuan: memperoleh kurva resistivitas semu (Ωm) dan fase (°) terhadap frekuensi (Hz), serta memeriksa kestabilan data.

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# 1. MEMBUAT DATA SINTETIS MEDAN LISTRIK DAN MAGNET
# =====================================================

np.random.seed(42)

frekuensi = np.logspace(-2, 2, 40)  # 0.01–100 Hz
omega = 2 * np.pi * frekuensi

# Model respons bawah permukaan sintetis
rho_model = 100 + 500 / (1 + (frekuensi / 1.0)**1.5)
fase_model = 45 + 15 * np.exp(
    -((np.log10(frekuensi) - 0.0)**2) / 1.5
)

# Impedansi sintetis: |Z| = sqrt(rho * omega * mu0)
mu0 = 4 * np.pi * 1e-7
Z_model = np.sqrt(rho_model * omega * mu0)

# Simulasi medan magnet dan medan listrik kompleks
H = np.ones(len(frekuensi), dtype=complex)
E = Z_model * np.exp(1j * np.deg2rad(fase_model)) * H

# Tambahkan noise pada data pengukuran sintetis
noise = np.random.normal(0, 0.08, len(frekuensi))
E_noisy = E * (1 + noise)

# Contoh gangguan pada beberapa frekuensi
indeks_noise = [8, 23, 31]
E_noisy[indeks_noise] *= 1.8

# =====================================================
# 2. MENGHITUNG RESISTIVITAS SEMU DAN FASE
# =====================================================

Z = E_noisy / H

rho_apparent = np.abs(Z)**2 / (omega * mu0)
fase = np.rad2deg(np.angle(Z))

# Tandai titik yang menyimpang dari kurva lokal
log_rho = np.log10(rho_apparent)
median_lokal = np.copy(log_rho)

for i in range(1, len(log_rho) - 1):
    median_lokal[i] = np.median(log_rho[i-1:i+2])

residual = np.abs(log_rho - median_lokal)
noise_flag = residual > 0.12

# =====================================================
# 3. SMOOTHING UNTUK PEMBANDINGAN
# =====================================================

def moving_average(data, window=3):
    hasil = np.copy(data)
    for i in range(len(data)):
        awal = max(0, i - window // 2)
        akhir = min(len(data), i + window // 2 + 1)
        hasil[i] = np.mean(data[awal:akhir])
    return hasil

rho_smooth = 10**moving_average(np.log10(rho_apparent))
fase_smooth = moving_average(fase)

# =====================================================
# 4. PEMERIKSAAN KUALITAS DATA
# =====================================================

print("ANALISIS MT - DATA SINTETIS")
print("-" * 45)
print("Jumlah titik frekuensi :", len(frekuensi))
print("Titik ditandai noise   :", np.sum(noise_flag))
print("Rentang frekuensi      :",
      f"{frekuensi.min():.3f}–{frekuensi.max():.1f} Hz")
print("Resistivitas minimum   :",
      f"{rho_apparent.min():.2f} ohm.m")
print("Resistivitas maksimum  :",
      f"{rho_apparent.max():.2f} ohm.m")

print("\nFrekuensi yang ditandai:")
for i in np.where(noise_flag)[0]:
    print(f"{frekuensi[i]:.3f} Hz")

# =====================================================
# 5. GRAFIK RESISTIVITAS SEMU
# =====================================================

plt.figure(figsize=(10, 5))

plt.loglog(frekuensi, rho_apparent, "o-",
           label="Data sebelum smoothing", alpha=0.7)
plt.loglog(frekuensi, rho_smooth, "-",
           label="Data setelah smoothing", linewidth=2)
plt.scatter(frekuensi[noise_flag],
            rho_apparent[noise_flag],
            color="red", marker="x", s=80,
            label="Titik terindikasi noise")

plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Resistivitas semu (Ωm)")
plt.title("Kurva Resistivitas Semu MT")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# =====================================================
# 6. GRAFIK FASE
# =====================================================

plt.figure(figsize=(10, 5))

plt.semilogx(frekuensi, fase, "o-",
             label="Data sebelum smoothing", alpha=0.7)
plt.semilogx(frekuensi, fase_smooth, "-",
             label="Data setelah smoothing", linewidth=2)
plt.scatter(frekuensi[noise_flag],
            fase[noise_flag],
            color="red", marker="x", s=80,
            label="Titik terindikasi noise")

plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Fase (derajat)")
plt.title("Kurva Fase MT")
plt.grid(True, which="both", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

## **4. Identifikasi Profil Kecepatan Gelombang Geser Menggunakan Metode MASW**

Studi bawah permukaan dilakukan untuk mengetahui variasi kecepatan gelombang geser terhadap kedalaman. Metode Multichannel Analysis of Surface Waves (MASW) digunakan dengan menganalisis gelombang permukaan yang direkam oleh beberapa geophone. Hasil survei digunakan untuk memperoleh profil Vs dan membantu mengetahui karakteristik material bawah permukaan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai metode Multichannel Analysis of Surface Waves (MASW).

T — Task: Identifikasi profil kecepatan gelombang geser (Vs) berdasarkan kurva dispersi data MASW.

C — Context: Suatu wilayah membutuhkan informasi mengenai variasi kecepatan gelombang geser terhadap kedalaman untuk mengetahui karakteristik bawah permukaan.

I — Input: Data rekaman seismik, posisi geophone, jarak antar-geophone, posisi sumber, dan waktu sampling.

K — Constraints: Jangan mengubah data rekaman asli dan pertahankan posisi serta jarak antar-geophone.

O — Output: Shot gather, kurva dispersi, dan profil Vs terhadap kedalaman.

Q — Quality Criteria: Kurva dispersi dapat dibaca dengan jelas, satuan konsisten, dan profil Vs dapat merepresentasikan data dispersi.

V — Optional Verification: Bandingkan kurva dispersi hasil pengolahan dengan data observasi dan periksa kesesuaian model Vs.

Data sintetis: simulasi rekaman seismik dari 24 geophone dengan jarak antar-geophone 2 m. Data digunakan untuk membuat shot gather, kurva dispersi, dan profil kecepatan gelombang geser (\(V_s\)) terhadap kedalaman.
Tujuan: memperlihatkan tahapan dasar analisis MASW untuk memperkirakan variasi kecepatan gelombang geser pada lapisan bawah permukaan.

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# =====================================================
# 1. PARAMETER SURVEI MASW
# =====================================================

n_geo = 24
dx = 2.0
pos_geo = np.arange(n_geo) * dx
dt = 0.002
durasi = 1.5

waktu = np.arange(0, durasi, dt)
frekuensi = np.linspace(5, 40, 100)

# Model Vs sintetis
kedalaman_model = np.array([0, 5, 12, 25])
Vs_model = np.array([180, 300, 500])

print("ANALISIS MASW - DATA SINTETIS")
print("Jumlah geophone :", n_geo)
print("Spasi geophone  :", dx, "m")
print("Interval waktu  :", dt, "s")
print("Model Vs        :", Vs_model, "m/s")

# =====================================================
# 2. MEMBUAT SHOT GATHER SINTETIS
# =====================================================

shot = np.zeros((n_geo, len(waktu)))

for i, x in enumerate(pos_geo):
    # Gelombang permukaan sintetis
    t_arrival = 0.15 + x / 250

    gelombang = np.sin(
        2 * np.pi * 12 * (waktu - t_arrival)
    ) * np.exp(
        -((waktu - t_arrival) / 0.12)**2
    )

    shot[i] = gelombang + np.random.normal(
        0, 0.025, len(waktu)
    )

# Plot shot gather
plt.figure(figsize=(11, 6))
plt.imshow(
    shot,
    aspect="auto",
    extent=[waktu[0], waktu[-1], pos_geo[-1], pos_geo[0]],
    cmap="seismic",
    interpolation="nearest"
)
plt.xlabel("Waktu (s)")
plt.ylabel("Posisi geophone (m)")
plt.title("Shot Gather MASW Sintetis")
plt.colorbar(label="Amplitudo relatif")
plt.tight_layout()
plt.show()

# =====================================================
# 3. KURVA DISPERSI SINTETIS
# =====================================================

# Frekuensi dan kecepatan fase ilustratif
f = frekuensi
c_fase = 180 + 320 * (f / 40)**0.8

# Simulasi energi dispersi
F, C = np.meshgrid(f, np.linspace(100, 600, 250))
energi = np.exp(
    -((C - (180 + 320 * (F / 40)**0.8)) / 18)**2
)

plt.figure(figsize=(10, 6))
plt.pcolormesh(
    F, C, energi,
    shading="auto",
    cmap="viridis"
)
plt.plot(
    f, c_fase,
    color="red",
    linewidth=2,
    label="Kurva dispersi ilustratif"
)
plt.xlabel("Frekuensi (Hz)")
plt.ylabel("Kecepatan fase (m/s)")
plt.title("Kurva Dispersi MASW Sintetis")
plt.colorbar(label="Energi relatif")
plt.legend()
plt.tight_layout()
plt.show()

# =====================================================
# 4. PROFIL Vs TERHADAP KEDALAMAN
# =====================================================

batas_kedalaman = np.array([0, 5, 12, 25])
Vs = np.array([180, 300, 500])

plt.figure(figsize=(6, 7))

for i in range(len(Vs)):
    plt.plot(
        [Vs[i], Vs[i]],
        [batas_kedalaman[i], batas_kedalaman[i+1]],
        color="blue",
        linewidth=3
    )

    if i < len(Vs) - 1:
        plt.plot(
            [Vs[i], Vs[i+1]],
            [batas_kedalaman[i+1],
             batas_kedalaman[i+1]],
            color="blue",
            linewidth=2
        )

plt.ylim(25, 0)
plt.xlim(0, 600)
plt.xlabel("Kecepatan gelombang geser Vs (m/s)")
plt.ylabel("Kedalaman (m)")
plt.title("Profil Vs Sintetis terhadap Kedalaman")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# =====================================================
# 5. RINGKASAN MODEL
# =====================================================

print("\nRINGKASAN MODEL LAPISAN")
for i in range(len(Vs)):
    print(
        f"Lapisan {i+1}: kedalaman "
        f"{batas_kedalaman[i]}–{batas_kedalaman[i+1]} m, "
        f"Vs = {Vs[i]} m/s"
    )
```

## **5. Pemeriksaan Kualitas Data Menggunakan Metode GPR**

Survei Ground Penetrating Radar (GPR) dilakukan untuk memperoleh informasi mengenai kondisi bawah permukaan dangkal. Sebelum dilakukan pengolahan lebih lanjut, data perlu diperiksa untuk mengetahui kelengkapan dan kualitas hasil akuisisi. Pemeriksaan dilakukan terhadap jumlah trace, interval sampling, frekuensi antena, dan posisi lintasan.

R — Role: Bertindaklah sebagai ahli geofisika yang menguasai pengolahan dan pemeriksaan data Ground Penetrating Radar (GPR).

T — Task: Periksa kualitas dan kelengkapan data GPR berdasarkan parameter akuisisi yang tersedia.

C — Context: Data GPR akan digunakan untuk analisis bawah permukaan sehingga perlu dipastikan memiliki kualitas dan kelengkapan yang memadai.

I — Input: Data trace GPR, interval sampling waktu, frekuensi antena, dan koordinat lintasan.

K — Constraints: Jangan mengubah raw trace, melakukan migrasi, atau menentukan velocity model.

O — Output: Laporan QC yang berisi jumlah trace, interval sampling, frekuensi antena, posisi, dan data yang bermasalah.

Q — Quality Criteria: Seluruh parameter utama tercatat, data bermasalah diberi flag, dan data asli tetap tidak berubah.

V — Optional Verification: Periksa kembali jumlah trace dan parameter sampling dengan metadata asli.

Data sintetis: simulasi data GPR dengan 200 trace, interval sampling 0,5 ns, frekuensi antena 200 MHz, dan posisi lintasan sepanjang 23 m.
Tujuan: memeriksa kelengkapan trace, interval sampling, frekuensi antena, koordinat lintasan, serta menandai data yang bermasalah tanpa mengubah data asli, melakukan migrasi, atau menentukan model kecepatan.

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# 1. PARAMETER AKUISISI GPR
# =====================================================

np.random.seed(42)

jumlah_trace = 200
interval_sampling_ns = 0.5
frekuensi_antena_mhz = 200
panjang_lintasan_m = 23.0

# Posisi trace sepanjang lintasan
posisi = np.linspace(0, panjang_lintasan_m, jumlah_trace)

# Data GPR sintetis: hanya untuk simulasi QC
jumlah_sampel = 512
raw_trace = np.random.normal(
    0, 1, (jumlah_trace, jumlah_sampel)
)

# Simulasikan trace bermasalah tanpa mengubah trace lainnya
raw_trace[30, :] = 0       # Trace kosong
raw_trace[100, :] = np.nan  # Trace mengandung nilai tidak valid

# =====================================================
# 2. PEMERIKSAAN KUALITAS DATA
# =====================================================

# Periksa trace kosong dan trace dengan nilai tidak valid
trace_kosong = np.all(
    np.nan_to_num(raw_trace) == 0, axis=1
)

trace_nan = np.any(
    ~np.isfinite(raw_trace), axis=1
)

# Periksa trace dengan amplitudo ekstrem
amplitudo_maks = np.full(jumlah_trace, np.nan)

for i in range(jumlah_trace):
    data = raw_trace[i]
    data_valid = data[np.isfinite(data)]

    if len(data_valid) > 0:
        amplitudo_maks[i] = np.max(np.abs(data_valid))

median_amp = np.nanmedian(amplitudo_maks)
mad_amp = np.nanmedian(
    np.abs(amplitudo_maks - median_amp)
)

if mad_amp > 0:
    trace_ekstrem = (
        np.abs(amplitudo_maks - median_amp) > 5 * 1.4826 * mad_amp
    )
else:
    trace_ekstrem = np.zeros(jumlah_trace, dtype=bool)

# Tandai trace yang bermasalah
flag_masalah = trace_kosong | trace_nan | trace_ekstrem

# =====================================================
# 3. LAPORAN QC
# =====================================================

print("=" * 50)
print("LAPORAN QUALITY CONTROL GPR")
print("=" * 50)
print("Jumlah trace             :", jumlah_trace)
print("Jumlah sampel per trace  :", jumlah_sampel)
print("Interval sampling        :", interval_sampling_ns, "ns")
print("Frekuensi antena         :", frekuensi_antena_mhz, "MHz")
print("Panjang lintasan         :", panjang_lintasan_m, "m")
print("Posisi awal              :", posisi[0], "m")
print("Posisi akhir             :", posisi[-1], "m")
print("Trace kosong             :", np.where(trace_kosong)[0] + 1)
print("Trace mengandung NaN/Inf :", np.where(trace_nan)[0] + 1)
print("Trace amplitudo ekstrem  :", np.where(trace_ekstrem)[0] + 1)
print("Total trace bermasalah   :", np.sum(flag_masalah))
print("=" * 50)

# =====================================================
# 4. VISUALISASI POSISI DAN FLAG QC
# =====================================================

plt.figure(figsize=(11, 4))

plt.scatter(
    posisi[~flag_masalah],
    np.zeros(np.sum(~flag_masalah)),
    color="green",
    label="Trace lolos pemeriksaan"
)

plt.scatter(
    posisi[flag_masalah],
    np.zeros(np.sum(flag_masalah)),
    color="red",
    marker="x",
    s=70,
    label="Trace bermasalah"
)

plt.xlabel("Posisi lintasan (m)")
plt.yticks([])
plt.title("Sebaran Posisi Trace dan Flag QC GPR")
plt.grid(True, axis="x", alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()

# =====================================================
# 5. VERIFIKASI DATA ASLI
# =====================================================

print("\nVERIFIKASI")
print("Bentuk array data       :", raw_trace.shape)
print("Jumlah posisi trace     :", len(posisi))
print("Jumlah trace terflag    :", np.sum(flag_masalah))
print("Data raw tidak diproses dengan filter atau migrasi.")
```
